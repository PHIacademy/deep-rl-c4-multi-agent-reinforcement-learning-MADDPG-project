````python
from unityagents import UnityEnvironment
import numpy as np

env = UnityEnvironment(file_name="/data/Tennis_Linux_NoVis/Tennis")
````

Environments contain **_brains_** which are responsible for deciding the actions of their associated agents. Here we check for the first brain available, and set it as the default brain we will be controlling from Python.

### 2. Examine the State and Action Spaces

Run the code cell below to print some information about the environment.

````python
# reset the environment
env_info = env.reset(train_mode=True)[brain_name]

# number of agents 
num_agents = len(env_info.agents)
print('Number of agents:', num_agents)

# size of each action
action_size = brain.vector_action_space_size
print('Size of each action:', action_size)

# examine the state space 
states = env_info.vector_observations
state_size = states.shape[1]
print('There are {} agents. Each observes a state with length: {}'.format(states.shape[0], state_size))
print('The state for the first agent looks like:', states[0])
````

### 3. Take Random Actions in the Environment

In the next code cell, you will learn how to use the Python API to control the agent and receive feedback from the environment.

Note that **in this coding environment, you will not be able to watch the agents while they are training**, and you should set `train_mode=True` to restart the environment.

````python
for i in range(5):                                         # play game for 5 episodes
    env_info = env.reset(train_mode=False)[brain_name]     # reset the environment    
    states = env_info.vector_observations                  # get the current state (for each agent)
    scores = np.zeros(num_agents)                          # initialize the score (for each agent)
    while True:
        actions = np.random.randn(num_agents, action_size) # select an action (for each agent)
        actions = np.clip(actions, -1, 1)                  # all actions between -1 and 1
        env_info = env.step(actions)[brain_name]           # send all actions to tne environment
        next_states = env_info.vector_observations         # get next state (for each agent)
        rewards = env_info.rewards                         # get reward (for each agent)
        dones = env_info.local_done                        # see if episode finished
        scores += env_info.rewards                         # update the score (for each agent)
        states = next_states                               # roll over states to next time step
        if np.any(dones):                                  # exit loop if episode finished
            break
    print('Total score (averaged over agents) this episode: {}'.format(np.mean(scores)))
````

### 5. Train the Agents with MADDPG

The cells below train the two agents using **Multi-Agent Deep Deterministic Policy Gradient (MADDPG)**.

Make sure the following three files are saved in the **same folder as this notebook** before running these cells:
- `hyperparameters.py` — all tunable constants (network sizes, learning rates, discount factor, noise, learning schedule, etc.)
- `model.py` — the `Actor` and `Critic` network definitions
- `maddpg_agents.py` — the `MADDPG` class (replay buffer, per-agent DDPG logic, centralized-critic learning update)


````python
import time
from collections import deque

import numpy as np
import torch
import matplotlib.pyplot as plt
%matplotlib inline

from maddpg_agents import MADDPG
from hyperparameters import SEED, NB_EPISODES, NB_STEPS, SOLVED_SCORE, SCORES_WINDOW
````

Re-open the environment in `train_mode=True` if you already closed it in the demonstration above (if it's still open, the cell below simply reuses it). We also run it with `no_graphics=True` since the agents can't be watched live in this workspace anyway, which speeds up training.

````python
try:
    env_info = env.reset(train_mode=True)[brain_name]
    print('Reusing existing environment.')
except Exception:
    from unityagents import UnityEnvironment
    env = UnityEnvironment(file_name="/data/Tennis_Linux_NoVis/Tennis")
    brain_name = env.brain_names[0]
    brain = env.brains[brain_name]
    env_info = env.reset(train_mode=True)[brain_name]

num_agents = len(env_info.agents)
state_size = env_info.vector_observations.shape[1]
action_size = brain.vector_action_space_size
print('Number of agents:', num_agents, '| State size:', state_size, '| Action size:', action_size)
````

````python
np.random.seed(SEED)
torch.manual_seed(SEED)

maddpg = MADDPG(num_agents=num_agents, state_size=state_size, action_size=action_size, seed=SEED)
````

The training loop below:
- resets the environment (`train_mode=True`) at the start of each episode
- lets both agents act (with exploration noise) and stores every transition in the shared replay buffer
- triggers MADDPG's learning update on the schedule defined in `hyperparameters.py`
- scores each episode as `max(agent_0_total_reward, agent_1_total_reward)`, per the project spec
- stops and saves a checkpoint as soon as the rolling 100-episode average reaches `SOLVED_SCORE` (0.5)

````python
def maddpg_train(env, brain_name, maddpg,
                  n_episodes=NB_EPISODES, max_t=NB_STEPS,
                  solved_score=SOLVED_SCORE, scores_window_size=SCORES_WINDOW,
                  checkpoint_prefix='checkpoint', print_every=100):
    """
    Train `maddpg` in `env` for up to `n_episodes` episodes of at most `max_t` steps each.

    Returns
    -------
    all_scores : list of the per-episode score (max over agents)
    avg_scores : list of the rolling `scores_window_size`-episode average, aligned with all_scores
    critic_losses : list of the (agent-averaged) critic loss recorded at the end of each
                    episode -- carries the previous value forward on episodes where no
                    learning update happened, so it's always aligned with all_scores
    actor_losses : same, for actor loss
    q_means : same, for the mean predicted Q-value (a rapidly growing magnitude here is
              the classic sign of Q-value divergence / overestimation)
    """
    scores_deque = deque(maxlen=scores_window_size)
    all_scores = []
    avg_scores = []
    critic_losses = []
    actor_losses = []
    q_means = []
    solved = False
    start_time = time.time()

    last_critic_loss, last_actor_loss, last_q_mean = None, None, None

    for i_episode in range(1, n_episodes + 1):
        env_info = env.reset(train_mode=True)[brain_name]
        states = env_info.vector_observations
        maddpg.reset_noise()
        episode_scores = np.zeros(maddpg.num_agents)

        for t in range(max_t):
            actions = maddpg.act(states, add_noise=True)
            env_info = env.step(actions)[brain_name]
            next_states = env_info.vector_observations
            rewards = np.array(env_info.rewards)
            dones = np.array(env_info.local_done)

            episode_done = bool(np.any(dones))
            maddpg.step(states, actions, rewards, next_states, dones, episode_done=episode_done)

            states = next_states
            episode_scores += rewards

            if episode_done:
                break

        episode_score = np.max(episode_scores)
        scores_deque.append(episode_score)
        all_scores.append(episode_score)
        avg_score = np.mean(scores_deque)
        avg_scores.append(avg_score)

        # Pull the latest training diagnostics (None until the first learn() call happens)
        c_loss, a_loss, q_mean = maddpg.get_diagnostics()
        if c_loss is not None:
            last_critic_loss, last_actor_loss, last_q_mean = c_loss, a_loss, q_mean
        critic_losses.append(last_critic_loss)
        actor_losses.append(last_actor_loss)
        q_means.append(last_q_mean)

        print('\rEpisode {}\tAverage Score: {:.3f}\tScore: {:.3f}\tCritic Loss: {}\tMean Q: {}'.format(
            i_episode, avg_score, episode_score,
            'n/a' if last_critic_loss is None else '{:.4f}'.format(last_critic_loss),
            'n/a' if last_q_mean is None else '{:.3f}'.format(last_q_mean)), end='')
        if i_episode % print_every == 0:
            elapsed = time.time() - start_time
            print('\rEpisode {}\tAverage Score: {:.3f}\tCritic Loss: {}\tMean Q: {}\tElapsed: {:.1f}s'.format(
                i_episode, avg_score,
                'n/a' if last_critic_loss is None else '{:.4f}'.format(last_critic_loss),
                'n/a' if last_q_mean is None else '{:.3f}'.format(last_q_mean),
                elapsed))

        if not solved and len(scores_deque) >= scores_window_size and avg_score >= solved_score:
            elapsed = time.time() - start_time
            print('\nEnvironment solved in {:d} episodes!\tAverage Score: {:.3f}\tElapsed: {:.1f}s'.format(
                i_episode, avg_score, elapsed))
            maddpg.save(checkpoint_prefix)
            solved = True
            break

    if not solved:
        print('\nTraining ended after {:d} episodes without reaching the target score.'.format(n_episodes))
        maddpg.save(checkpoint_prefix + '_unsolved')

    return all_scores, avg_scores, critic_losses, actor_losses, q_means
````

````python
scores, avg_scores, critic_losses, actor_losses, q_means = maddpg_train(env, brain_name, maddpg)
````

Plot the score of each episode together with the rolling 100-episode average.

````python
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111)
plt.plot(np.arange(1, len(scores) + 1), scores, alpha=0.4, label='Score per episode')
plt.plot(np.arange(1, len(avg_scores) + 1), avg_scores, label='Rolling 100-episode average')
plt.axhline(y=SOLVED_SCORE, color='r', linestyle='--', label='Solved threshold (0.5)')
plt.ylabel('Score')
plt.xlabel('Episode #')
plt.legend()
plt.show()
````

### 7. Evaluate the Trained Agent

Load the saved checkpoint weights and run a few evaluation episodes (no exploration noise) to check the final policy's performance. `MADDPG.load()` restores all four files (`checkpoint_actor_agent0.pth`, `checkpoint_actor_agent1.pth`, `checkpoint_critic_agent0.pth`, `checkpoint_critic_agent1.pth`) in one call.

````python
# Load the saved weights for both agents (actor + critic, local nets;
# targets are hard-synced to match inside load())
maddpg.load(path_prefix='checkpoint')

n_eval_episodes = 5
eval_scores = []

for i_episode in range(1, n_eval_episodes + 1):
    env_info = env.reset(train_mode=False)[brain_name]  # reset the environment
    states = env_info.vector_observations                # shape: (num_agents, state_size)
    episode_scores = np.zeros(maddpg.num_agents)

    while True:
        actions = maddpg.act(states, add_noise=False)     # no exploration noise -- pure policy
        env_info = env.step(actions)[brain_name]
        next_states = env_info.vector_observations
        rewards = np.array(env_info.rewards)
        dones = np.array(env_info.local_done)

        episode_scores += rewards
        states = next_states

        if np.any(dones):
            break

    episode_score = np.max(episode_scores)  # same scoring convention used during training
    eval_scores.append(episode_score)
    print('Episode {}\tScore: {:.3f}'.format(i_episode, episode_score))

print('\nAverage score over {} evaluation episodes: {:.3f}'.format(n_eval_episodes, np.mean(eval_scores)))
````
